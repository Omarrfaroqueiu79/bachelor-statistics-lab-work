# Example 3:
Write an R function to implement exercise 1 where p and random seed are parameters.

```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
def random_Walk1(prob1,prob2,s):
    random.seed(s)
    random.seed(10)
    xt = random.choices([-1,1],weights=[prob1,prob2],k=20)
    Sn = 5+np.cumsum(xt)
    n = np.arange(0,len(xt))
    plt.plot(n,Sn,color="blue")
    plt.xlabel("n")
    plt.ylabel("Sn")
    plt.title(f"Random walk for probabilities {prob1,prob2}")
    plt.show()
```

```python
random_Walk1(0.3,0.1,1000)
```
