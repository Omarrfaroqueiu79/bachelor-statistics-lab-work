## generate first the [-1,1] values

```python
import random
        
```

### generated the random numbers

```python
ls = [-1,1]
x1 = random.choices(ls,k=20)
x1
```

## cumulative sum

```python
import numpy as np
```

```python
Sn=1+np.cumsum(x1)
Sn
```

```python
import matplotlib.pyplot as plt
import seaborn
import pandas
```

```python
n = np.arange(0,len(x1))
n
```

```python
plt.plot(n,Sn,linestyle = 'solid')
plt.show()
```
