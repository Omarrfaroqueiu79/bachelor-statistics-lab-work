# Example 1
A player flips coin to mimic what he can gain after certain numbers of playing. If the coin
showes head, he will gain 1 pound otherwise he will lose 1 pound. Assume he has 5 pounds at begining and
please generate a simple random walk process to show hwo much money the play will have after 20 plays and
plot this process again the steps.We assume that the chance of the player get a head and a tail when he flips
a corn are:
1. prob=c(0.5, 0.5)
2. prob=c(0.2, 0.8)
3. prob=c(0.7, 0.3)

```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

## for prob=[0.5,0.5]

```python
x = random.choices([-1,1],(0.5,0.5),k=20)
x
```

```python
Sn = 5+np.cumsum(x)
Sn
```

```python
n = np.arange(0,len(Sn))
n
```

```python
plt.plot(n,Sn,color="blue")
plt.title("Random Walk for prob(0.5.5)")
plt.xlabel("n")
plt.ylabel("Sn")
plt.show()
```

### for prob =[0.2,0.8]

```python
x = random.choices([-1,1],weights=[0.2,0.8],k=20)
x
```

```python
Sn = 5+np.cumsum(x)
```

```python
n = np.arange(0,len(x))
```

```python
plt.plot(n,Sn,color="red")
plt.title("Random Walk for prob(0.2.0.8)")
plt.xlabel("n")
plt.ylabel("Sn")
plt.show()
```

### for prob =[0.7,0.3]

```python
x = random.choices([-1,1],weights=[0.7,0.3],k=20)
x
```

```python
Sn = 5+np.cumsum(x)
Sn
```

```python
n = np.arange(0,len(x))
```

```python
plt.plot(n,Sn,color="green")
plt.xlabel("n")
plt.ylabel("Sn")
plt.title("Random Walk for prob(0.7,0.3)")
plt.show()
```
