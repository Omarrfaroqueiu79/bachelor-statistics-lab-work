```python
import random
import numpy as np
import scipy as sp
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('soft_drink.csv')
df.head()
```

## Current State and Next State

```python
cstate = df['Current State']
nstate = df['Next State']
```

```python
tpm = pd.crosstab(cstate,nstate)
tpm
```

```python
tpm = np.array(tpm)
tpm
```

```python
rowSums = np.sum(tpm,1)
rowSums
```

```python
TPM = tpm.transpose()/rowSums
TPM = TPM.transpose()
```

# Random Walk on Markov Chain
### for 15 Steps

```python
states = ["Coca Cola","Pepsi","RC","Sprite"]
n = 15
start_state = 0 # starting with coacola
print(states[start_state],"-->",end=" ")
prev_state = start_state
for i in range(0,n):
    curr_state = np.random.choice([0,1,2,3],p=TPM[prev_state])
    print(states[curr_state],"-->",end=" ")
    prev_state = curr_state
print("Stop")
```

### We can perform the random walk for different states

# Stationary For Transition Probability Matrix

```python
steps = 10**6
start_state = 0
pi = np.array([0,0,0,0])
pi[start_state]=1
prev_state = start_state
i = 0
while i<steps:
    curr_state = np.random.choice([0,1,2,3],p=TPM[prev_state])
    pi[curr_state]+=1
    i+=1
pi = pi/steps
pi
```
