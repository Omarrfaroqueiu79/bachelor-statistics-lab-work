# Random Walk

```python
import random
import numpy as np
```

```python
tpm = np.array([[0.31111111, 0.26666667, 0.06666667, 0.35555556],
       [1.        , 0.        , 0.        , 0.        ],
       [0.5       , 0.        , 0.5       , 0.        ],
       [0.58333333, 0.16666667, 0.16666667, 0.08333333]])
```

```python
tpm
```

```python
states = ["Coca Cola","Pepsi","RC","Sprite"]
```

```python
n = 15
start_state = 0 # starting with coacola
print(states[start_state],"-->",end=" ")
prev_state = start_state
for i in range(0,n):
    curr_state = np.random.choice([0,1,2,3],p=tpm[prev_state])
    print(states[curr_state],"-->",end=" ")
    prev_state = curr_state
print("Stop")
```

```python
n = 15
start_state = 1 # starting with Pepsi
print(states[start_state],"-->",end=" ")
prev_state = start_state
for i in range(0,n):
    curr_state = np.random.choice([0,1,2,3],p=tpm[prev_state])
    print(states[curr_state],"-->",end=" ")
    prev_state = curr_state
print("Stop")
```

# Monte Carlo Approach 1

```python
steps = 10**6
start_state = 0
pi = np.array([0,0,0,0])
pi[start_state]=1
prev_state = start_state
i = 0
while i<steps:
    curr_state = np.random.choice([0,1,2,3],p=tpm[prev_state])
    pi[curr_state]+=1
    i+=1
pi
```

```python
pi/steps
```

# Repeated Matrix Multiplication 
### its better


```python
steps = 10**6
tem = tpm
i =0
while i<steps:
    tem = np.matmul(tem,tpm)
    i +=1
tem
```
