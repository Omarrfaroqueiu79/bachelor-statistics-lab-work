# Example
Test the hypothesis that the median is at least 42 years old

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as st
```

```python
x = np.array([56,62,61,54,52,32,24,52,49,26,31,31,54,38,36,45,53,37,40,38,31,29,25,45,32,49,39,36,38])
```

```python
x.size
```

```python
len(x)
```

```python
mu0 = 42
```

```python
signed = x-mu0
```

```python
signed
```

```python
y = []
for i in signed:
    if i>0:
        y.append(1)
    else:
        y.append(0)
```

```python
y = np.array(y)
y.size
```

```python
y
```

```python
Tpos = sum(y==1)
Tpos
```

```python
Tneg = sum(y==0)
Tneg
```

```python
T = min(Tpos,Tneg)
T
```

```python
n = Tpos+Tneg
```

```python
from math import sqrt
```

```python
n
```

```python
p=.5
q=.5
m = n*p
sd = sqrt(n*p*q)
```

```python
if Tpos<m:
    z = (Tpos+.5-m)/sd
else:
    z = (Tpos-.5-m)/sd
```

```python
z
```

### Z critical value for 5% level of significance

```python
import scipy.stats as st
```

```python
z_c=st.norm.ppf(.05)
```

```python
if z<z_c:
    print("We may reject H0")
else:
    print("We may fail to rejet H0")
```

### P value approach
if p < .05 reject H0 otherwise accept H0

```python
p = st.norm.cdf(z)
```

```python
p
```

```python
if p<.05:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```
