## Large Sample
---

```python
x = np.array([56,62,61,54,52,32,24,35,50,42,52,49,26,31,31,54,38,36,45,53,37,40,38,31,29,25,45,32,49,39,36,38])
```

```python
print(x)
print(len(x))
```

```python
m0 = 42
s = x-m0
s
```

```python
po = len(s[s>0])
neg = len(s[s<0])
print(po,neg)
```

```python
n = po+neg
```

```python
n
```

### As n =31 it is a large sample problem

```python
if po < n/2:
    z = ((po+0.5)-(n/2))/(.5*np.sqrt(n))
else:
    z = ((po-0.5)-(n/2))/(0.5*np.sqrt(n))
z
```
