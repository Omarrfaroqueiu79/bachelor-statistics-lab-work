# Markov Chain

```python
import numpy as np
```

```python
states = {0: "Burger",
          1: "Pizza",
          2: "Hotdog"
         }
```

## Transition Matrix

```python
A = np.array([[.2,.6,.2],[.3,.0,.7],[.5,0,.5]])
A
```

## Simulating a random walk

```python
n = 15 # number of transition
start_states = 0
print(states[start_states],"-->",end=" ")
prev_state = start_states

while n-1:
    curr_state = np.random.choice([0,1,2],p=A[prev_state])
    print(states[curr_state],"-->",end=" ")
    prev_state = curr_state
    n-=1
print("stop")
```

## Approach 1: Monte Carlo

```python
steps = 10**6
start_state = 0
pi = np.array([0,0,0])
pi[start_state]=1
prev_state = start_state

i=0
while i<steps:
    curr_state = np.random.choice([0,1,2],p=A[prev_state])
    pi[curr_state]=1
    prev_state = curr_state
    i += 1
print(" pi = ",pi/steps)
```

## Approach 2: Repeated Matrix Multiplication

```python
steps = 10**3
A_n = A
i = 0
while i<steps:
    A_n = np.matmul(A_n,A)
    i+=1
print("A^n=\n",A_n,"\n")
print("pi = ",A_n[0])
```

```python
steps = 10**6 # increase the steps to increase the accuracy
A_n = A
i = 0
while i<steps:
    A_n = np.matmul(A_n,A)
    i+=1
print("A^n=\n",A_n,"\n")
print("pi = ",A_n[0])
```

## Approach 3: Finding left Eigen Vectors

```python
import scipy.linalg
values,left = scipy.linalg.eig(A,right = False,left = True)
print("Left eigen vectors= \n",left,"\n")
print("eigen values = \n",values)
```

```python
pi = left[:,0]
pi_normalized = [(x/np.sum(pi)).real for x in pi]
pi_normalized
```

# P(pizza--> Hotdog--> Hotdog --> Burger)=?
## => p(x_0=pizza,x_1=Hotdog,X_2=Hotdog,x_3 =Burger)
## => p(x_0=pizza) p(x_1=Hotdog | x_0 = pizza) P(x_2 = hotdog|x_1 = hotdog) P(x_2=Burger|x_2=Hotdog)

```python
def find_prob(seq,A,pi):
    start_state = seq[0]
    prob= pi[start_state]
    prev_state = start_state
    for i in range(1,len(seq)):
        curr_state = seq[i]
        prob *= A[prev_state][curr_state]
        prev_state = curr_state
    return prob
print(find_prob([1,2,2,0],A,pi_normalized))
```
