# Example 2:
Write an R function to implement exercise 1 where the probability of prob = (p, q), where
p = P (Xt = 1), q = P (Xt = −1) is used as parameter of the function. We give the name of function as
randomwalk1 and the R code is as following

```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
def randomWalk1(prob1,prob2):
    random.seed(10)
    xt = random.choices([-1,1],weights=[prob1,prob2],k=20)
    Sn = 5+np.cumsum(xt)
    n = np.arange(0,len(xt))
    plt.plot(n,Sn,color="blue")
    plt.xlabel("n")
    plt.ylabel("Sn")
    plt.title(f"Random walk for probabilities {prob1,prob2}")
    plt.show()
```

```python
randomWalk1(0.3,0.7)
```

```python
randomWalk1(0.5,0.5)
```

```python
randomWalk1(0.7,0.3)
```
