```python
males = [19, 22, 16, 29, 24]
females = [20, 11, 17, 12]
```

```python
from scipy.stats import mannwhitneyu
```

```python
U1, p = mannwhitneyu(males, females, method="exact")
```

```python
U1,p
```

### P-value = o.111 
We may accept our null hypothesis. alpha = .5

```python
# U1 = the test statistic for male
ux,uy = len(males),len(females)
ux,uy
```

```python
U2 = ux*uy-U1
U2
```

```python
pnorm = mannwhitneyu(males,females,method="asymptotic")
pnorm
```

```python
U = min(U1,U2)
N = ux+uy
z = (U-(ux*uy)/1+.5)/np.sqrt(ux*uy*(N+1)/12)
z
```

```python
p = 2*norm.cdf(z)
```
