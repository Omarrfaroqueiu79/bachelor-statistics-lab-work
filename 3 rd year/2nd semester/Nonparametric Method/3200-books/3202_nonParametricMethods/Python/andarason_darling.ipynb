# Mannwhitney U Test


```python
males = [19, 22, 16, 29, 24]
females = [20, 11, 17, 12]
males,females
```

```python
u1,p =scipy.stats.mannwhitneyu(males,females,method='exact')
u1
```

```python
u1,p
```

## Since p>0.05, predefined level of significance we may reject our null hypothesis and conclude that there are significnt difference in male and female
