# Wilcoxon rank test for median

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import*
```

```python
x = np.array([
    9.4,13.4,15.6,16.2,16.4,16.8,18.1,18.7,18.9,19.1,19.3,20.1,20.4,21.6,21.9,23.4,23.5,24.8,24.9,26.8
])
x
```

```python
len(x)
```

```python
res = wilcoxon(x)
```

```python
res
```

```python
res.statistic,res.pvalue
```

```python
x2 = np.array([6, 8, 14, 16, 23, 24, 28, 29, 41, -48, 49, 56, 60, -67, 75])
```

```python
res2=wilcoxon(x2)
```

```python
res2.statistic,res2.pvalue
```

```python
np.quantile(x2,.25),np.quantile(x2,.5),np.quantile(x2,.75)
```

```python
res3 = wilcoxon(x,method="approx")
res3.statistic,res3.pvalue
```

```python
res4 = wilcoxon(x,alternative='greater')
res4.statistic,res4.pvalue
```
