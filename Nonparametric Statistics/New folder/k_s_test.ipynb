## Kolmogrov-smirnov one sample test



```python
import numpy as np 
#random number generate from normal distribution
```

```python
f_x=[0.32,0.11,0.44,0.51,0.53,0.57,0.60,0.63,0.65,0.69,0.72,0.76,0.79,0.81,0.83,0.87,0.91,0.94,0.96,0.98]
```

```python
f_x.sort()
print(f_x)
# print(f_x_sorted)
N=len(f_x)
print(N)
# f_x_sorted
```

```python
# find the maximum value(i/N-Ri)
plus_max=list()
for i in range(1,N+1):
    x=(i/N)-f_x[i-1]
    plus_max.append(x)

k_plus_max=np.sqrt(N)*np.max(plus_max)
k_plus_max
```

```python
# find the d minus( Ri-(i-1)/N)

minus_max=list()
for i in range(1,N+1):
    y=(i-1)/N
    x=f_x[i-1]-y
    minus_max.append(x)

k_minus_max=np.sqrt(N)*np.max(minus_max)
k_minus_max
```

```python
K_max=max(k_plus_max,k_minus_max)
K_max
```
