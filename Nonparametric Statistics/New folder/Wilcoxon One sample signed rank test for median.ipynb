## The wilcoxon signed rank test for median for one sample


```python
# QUESTION

let denote the length, in centimeters, of a randomly selected pygmy sunfish, 
. If we obtain the following data set:

 5.0 3.9 5.2 5.5 2.8 6.1 6.4 2.6 1.7 4.3

can we conclude that the median length of pygmy sunfish differs significantly from 3.7 centimeters?
```

```python
l=[5.0, 3.9, 5.2, 5.5, 2.8, 6.1, 6.4 ,2.6 ,1.7, 4.3]
```

```python
#Hypothesis:
#Null: H0=3.7
#Alternative: H1:=! 3.7
```

```python
from scipy.stats import wilcoxon
import statistics as stat
```

```python
l_median=stat.median(l)
```

```python
l_median
```

```python
result=wilcoxon(l)
```

```python
result.statistic,result.pvalue
```

```python
res=wilcoxon(l,alternative='greater')
res.statistic,res.pvalue
```

```python
# Comment: the null hypothesis is rejected.that means the median (3.7) is differ.
```
