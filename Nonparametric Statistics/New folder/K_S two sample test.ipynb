## Kolmogrove smirnov two sample test

```python
from scipy import stats
```

```python
x=[1.41,0.26,1.97, 0.33, 0.55, 0.77, 1.46, 1.18,1.20,.50]
y=[108 ,112, 117, 130 ,111 ,131 ,113, 113 ,105 ,128]
```

```python
## hypothesis:
#null: the two sample come from same distribution.
# alternative: samples are from differnt distribution.
```

```python
result=stats.kstest(x,y)
result
```

```python
#Here p_value is greater than the leval of significance(0.05).so there is no
# enougth information to rejecet the null hypothesis .
# so the two distribution comes from two different distribution.
```
