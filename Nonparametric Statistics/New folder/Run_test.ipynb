## Run test-test of randomness

```python
####################-------------run test----------###############
import numpy as np
import math
import statistics
```

```python
# #---------------creating the run test function---------------###########
# def runsTest(l,l_median):
#     runs,n1,n2=0,0,0
```

```python
l=[16.75,19.25,22.00,20.50,15.50,17.25,20.75,24.75,21.50,19.75,17.50,22.75,23.50,13.00,19.00]
```

```python
l_median=statistics.median(l)
```

```python
#---------------creating the run test function---------------###########
def runsTest(l,l_median):
    runs,n1,n2=0,0,0


    for i in range(len(l)):
        if(l[i]>=l_median and l[i-1]<l_median) or \
            (l[i]<l_median and l[i-1]>=l_median):
            runs +=1
        #number of positives values
        if(l[i])>=l_median:
            n1 +=1
        else:
            n2 +=1
    runs_exp = ((2*n1*n2)/(n1+n2))+1
    stan_dev = math.sqrt((2*n1*n2*(2*n1*n2-n1-n2))/(((n1+n2)**2)*(n1+n2-1)))  
                       
    z = (runs-runs_exp)/stan_dev 
    return z 
```

```python
z=abs(runsTest(l, l_median))
```

```python
z
```

```python
# we know that the  z value at the 5% level of significance is 1.96
if z<=1.96:
    print("The run is not statistically significant")
else:
    print("The test is significant")
```
