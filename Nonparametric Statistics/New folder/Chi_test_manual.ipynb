## Chi square test

```python
import math
import numpy as np
import statistics
import scipy.stats as stats
```

```python
x=[224,119,130,48,59]
p=[0.4,0.2,0.2,0.1,0.1]
```

```python
sum_x=sum(x)
print(sum_x)
```

```python
for i in p:
    y=sum_x*i
    print(y)
    
   
```

```python
expected=[232,116,116,58,58]
```

```python
chi_square_test_statistic=0
for i in range(len(x)):
    chi_square_test_statistic=chi_square_test_statistic +\
        (np.square(x[i]-expected[i]))/expected[i]

print('chi square value determined by formula:'+
      str(chi_square_test_statistic))

    
```

```python
#--------find chi_square_ critical value-------
print(stats.chi2.ppf(1-0.05,df=4))
```

```python
p_value=stats.chi2.pdf(3.79,df=4)
print(p_value)
```

```python
alpha=0.05
if p_value>=alpha:
    print('Independent-Ho accept')
else:
    print('Dependent-Ho reject')
```
