####The Wilcoxon signed-rank test is the non-parametric univariate test which is an alternative to the dependent t-test. It also is called the Wilcoxon T test, most commonly so when the statistic value is reported as a T value.This is the recommended test to use when the data violates the assumption of normality. It’s used to test if there is a significant difference on scores when there is a “before” and “after” condition of some treatment or intervention.

```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("blood_pressure.csv")
```

```python
#we want to justify the blood presure before and after treatment.
df.head()
```

```python
df[["bp_before","bp_after"]].describe()
```

```python
plt.hist(df["bp_before"],color="green")
plt.title("Histogran_bp_before")
```

```python
plt.hist(df["bp_after"],color="red")
plt.title("BP_AFTER")
```

```python
stats.probplot(df["bp_before"],dist="norm",plot=plt)
plt.title("Blood Pressure Before Q-Q Plot")
plt.savefig("BP_before_Q-Q.png")
```

```python
stats.probplot(df["bp_after"],dist="norm",plot=plt)
plt.title("Blood Pressure after Q-Q Plot")
plt.savefig("BP_after_Q-Q.png")
```

## Let’s test this statistically to see if the data is normally distributed. To test this, we use the Shapiro-Wilk test for normality

```python
stats.shapiro(df["bp_before"])
```

```python
stats.shapiro(df["bp_after"])
```

```python
###Both of the variables have a significant p-value which means each variable violates the assumption of normality.
# Therefore, the Wilcoxon signed rank test
```

```python
stats.wilcoxon(df["bp_before"],df["bp_after"])
```

### Comeent:A Wilcoxon T test was used to analyze the blood pressure before and after the intervention to test if the intervention had a significant affect on the blood pressure.
