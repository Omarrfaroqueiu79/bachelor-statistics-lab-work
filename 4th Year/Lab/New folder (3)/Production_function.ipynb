```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
```

```python
df=pd.read_csv("Production_function_7_3(1).csv")
df.isnull()
```

```python
df.head()
```

```python
# scatter plot for normal data
plt.scatter(df.X2,df.Y,color='green')
plt.scatter(df.X3,df.Y,color="red")
plt.ylabel('Output')
plt.title('Production function')
plt.legend(['Green labour','Red capital'],loc='lower right')
plt.show()
```

```python
plt.plot(df.X2)
plt.plot(df.X3)

plt.plot(df.lnX2,color='0')
plt.plot(df.lnX3,color='red')
```

```python
x=df[['X2','X3']]
from sklearn import linear_model
model=linear_model.LinearRegression()
model.fit(x,df.Y)
```

```python
r_square=model.score(x,df.Y)
print('Coefficient of determination :',r_square)
```

```python
print('coefficient: ',model.coef_)
```

```python
# log linear model fitting
lnx=df[['lnX2','lnX3']]
log_model=linear_model.LinearRegression()
log_model.fit(lnx,df.lnY)
```

```python
r_square=log_model.score(lnx,df.lnY)
print('Coefficient of determination :',r_square)
```

```python
print('coefficient: ',log_model.coef_)
```
