## Chow test the structural change test

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
```

```python
df=pd.read_csv("chow_test.csv")
```

```python
df.tail(5)
```

```python
# Plotting the data set
plt.scatter(df.Income,df.Savings,color='m')
plt.xlabel("Income")
plt.ylabel("Savings")
plt.plot(df.Income,df.Savings)
plt.title("Income vs Savings structural change")
plt.show()
```

```python
# performing the chow test
# import chowtest
plt.xlabel("Year")
plt.ylabel("Savings")
plt.plot(df.Year,df.Savings)
plt.title("Year vs Savings structural change")
plt.show()
```

```python
import statsmodels.api as sm
reg=sm.OLS(df.Income,df.Savings).fit()
ssr_total=reg.ssr
print('SSR 1970-1995:',reg.ssr)
```

```python
sav=df.Savings
inc=df.Income
sav_before=sav[:13]
inc_before=inc[:13]
sav_after=sav[13:]
inc_after=inc[13:]
print(inc_before)
```

```python
# structural change shows by the scatter plot
plt.scatter(inc_before,sav_before,color="yellow")
plt.plot(inc_before,sav_before,color='green')
plt.scatter(inc_after,sav_after,color='red')
plt.plot(inc_after,sav_after,color='m')
```

```python
reg_before=sm.OLS(inc_before,sav_before).fit()
reg_after=sm.OLS(inc_after,sav_after).fit()
ssr_before=reg_before.ssr
ssr_after=reg_after.ssr
print("SSR 1970-1981 :",reg_before.ssr)
print("SSR 1982-1995 :",reg_after.ssr)
```

```python
# # compute the F-distributed Chow test statistic
# Compute the numerator 

# k=2 degrees of freedom=2 OLS coefficients alpha and beta
# *compute the denominator
#   22 degrees of freedom =total number of data points(26)-2*k

numerator=(ssr_total-(ssr_before+ssr_after))/2
denominator=(ssr_before+ssr_after)/22

chow_test=numerator/denominator
print('Chow test statistic',chow_test)
```
