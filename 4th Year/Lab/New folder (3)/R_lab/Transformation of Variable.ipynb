```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

```python
import warnings
warnings.filterwarnings("ignore")
```

#------------------------------------------------------------------------------------------------------------#

```python
x = np.random.poisson(50,1000000)
```

```python
x[:100]
```

### Centering

```python
x_centre = x-x.mean()
x_centre[:10]
```

### Auto Scaling

```python
x_auto = (x-x.mean())/x.std()
```

### Range Scaling

```python
x_range = (x-x.mean())/(x.max()-x.min())
```

### Normalization

```python
x_norma = (x-x.min())/(x.max()-x.min())
```

### Pareto Scaling

```python
x_pareto = (x-x.mean())/np.sqrt(x.std())
```

### Vast Scaling

```python
x_vast = ((x-x.mean())/x.std() )* (x.mean()/x.std())
```

### Level Scaling

```python
x_level = (x-x.mean())/x.mean()
```

### Log Transformation

```python
x_log10 = np.log10(x)
```

### Power Transformation

```python
x_pow = np.power(x,1/len(x))
```

### Box-Cox Transformation
delta = 4

```python
delta = 4
x_box_cox = (np.power(x,delta)-1)/delta
```

```python
transformed_data = [x_centre,x_auto,x_range,x_norma,x_pareto,
                    x_vast,x_level,x_log10,x_pow,x_box_cox]
titles = ["x_centre","x_auto","x_range","x_norma","x_pareto","x_vast","x_level","x_log10",
          "x_pow","x_box_cox"]
```

## Figures

```python

l = len(transformed_data)
for i in range(l):
    plt.figure(figsize=(8,4))
    plt.title(f"Original and {titles[i]}")
    plt.plot(x[:200],color = "blue")
    plt.plot(transformed_data[i][:200],color="green")
    plt.legend('Generated',"Transformed")
```
