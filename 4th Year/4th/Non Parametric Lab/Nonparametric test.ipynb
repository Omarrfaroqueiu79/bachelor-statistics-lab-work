## Chi-Square test

```python
import math
import numpy as np
import statistics
import scipy.stats as stats
```

```python
x=[224,119,130,48,59]
p=[0.4,0.2,0.2,0.1,0.1]
```

```python
sum_x=sum(x)
print(sum_x)
```

```python
for i in p:
    y=sum_x*i
    print(y)
    
   
```

```python
expected=[232,116,116,58,58]
```

```python
chi_square_test_statistic=0
for i in range(len(x)):
    chi_square_test_statistic=chi_square_test_statistic +\
        (np.square(x[i]-expected[i]))/expected[i]

print('chi square value determined by formula:'+
      str(chi_square_test_statistic))

    
```

```python
#--------find chi_square_ critical value-------
print(stats.chi2.ppf(1-0.05,df=4))
```

```python
p_value=stats.chi2.pdf(9.49,df=4)
print(p_value)
```

```python
alpha=0.05
if p_value>=alpha:
    print('Independent-Ho accept')
else:
    print('Dependent-Ho reject')
```

# Kolmogrov-smirnov one sample test

```python
import numpy as np 
#random number generate from normal distribution
```

```python
f_x=[108,112,117,130,111,131,113,113,105,128 ]
```

```python
f_x.sort()
print(f_x)
# print(f_x_sorted)
N=len(f_x)
print(N)
# f_x_sorted
```

```python
# find the maximum value(i/N-Ri)
plus_max=list()
for i in range(1,N+1):
    x=(i/N)-f_x[i-1]
    plus_max.append(x)

k_plus_max=np.sqrt(N)*np.max(plus_max)
k_plus_max
```

```python
# find the d minus( Ri-(i-1)/N)

minus_max=list()
for i in range(1,N+1):
    y=(i-1)/N
    x=f_x[i-1]-y
    minus_max.append(x)

k_minus_max=np.sqrt(N)*np.max(minus_max)
k_minus_max
```

```python
K_max=max(k_plus_max,k_minus_max)
K_max
```

# Test of randomness (Run test)

```python
####################-------------run test----------###############
import numpy as np
import math
import statistics
```

```python
# #---------------creating the run test function---------------###########
# def runsTest(l,l_median):
#     runs,n1,n2=0,0,0
```

```python
l=[27, 6, 19, 24, 18, 12, 15, 17, 18, 20,0, 9, 4, 12, 3, 2, 7, 7, 0, 5, 32, 16, 38, 31, 27, 15, 5, 9, 4, 10 ]
```

```python
l_median=statistics.median(l)
print(l_median)
```

```python
#---------------creating the run test function---------------###########
def runsTest(l,l_median):
    runs,n1,n2=0,0,0


    for i in range(len(l)):
        if(l[i]>=l_median and l[i-1]<l_median) or \
            (l[i]<l_median and l[i-1]>=l_median):
            runs +=1
        #number of positives values
        if(l[i])>=l_median:
            n1 +=1
        else:
            n2 +=1
    runs_exp = ((2*n1*n2)/(n1+n2))+1
    stan_dev = math.sqrt((2*n1*n2*(2*n1*n2-n1-n2))/(((n1+n2)**2)*(n1+n2-1)))  
                       
    z = (runs-runs_exp)/stan_dev 
    return z 
```

```python
z=abs(runsTest(l, l_median))
print(z)
```

```python
# we know that the  z value at the 5% level of significance is 1.96
if z<=1.96:
    print("The run is not statistically significant")
else:
    print("The test is significant")
```

# Wilcoxon sign rank test

```python
# The Wilcoxon signed-rank test is the non-parametric univariate test which is an alternative to the dependent t-test.
# It also is called the Wilcoxon T test, most commonly so when the statistic value is reported as a T value.
# This is the recommended test to use when the data violates the assumption of normality.
# It’s used to test if there is a significant difference on scores when there is a “before” and “after” condition of some treatment or intervention.
```

```python
## hypothesis:
# Null hypothesis (H0): The difference between the pairs follows a symmetric distribution around zero.
# Alternative hypothesis (HA): The difference between the pairs does not follow a symmetric distribution around zero.
```

```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("blood_pressure.csv")
```

```python
#we want to justify the blood presure before and after treatment.
df.head()
```

```python
df[["bp_before","bp_after"]].describe()
```

```python
plt.hist(df["bp_before"],color="green")
plt.title("Histogran_bp_before")
```

```python
plt.hist(df["bp_after"],color="red")
plt.title("BP_AFTER")
```

```python
stats.probplot(df["bp_before"],dist="norm",plot=plt)
plt.title("Blood Pressure Before Q-Q Plot")
# plt.savefig("BP_before_Q-Q.png")
```

```python
stats.probplot(df["bp_after"],dist="norm",plot=plt)
plt.title("Blood Pressure after Q-Q Plot")
# plt.savefig("BP_after_Q-Q.png")
```

## Let’s test this statistically to see if the data is normally distributed. To test this, we use the Shapiro-Wilk test for normality

```python
stats.shapiro(df["bp_before"])
```

```python
stats.shapiro(df["bp_after"])
```

```python
###Both of the variables have a significant p-value which means each variable violates the assumption of normality.
# Therefore, the Wilcoxon signed rank test
```

```python
stats.wilcoxon(df["bp_before"],df["bp_after"])
```

# Wilcoxon sign test for median

```python
# QUESTION

# let denote the length, in centimeters, of a randomly selected pygmy sunfish, 
# . If we obtain the following data set:

#  5.0 3.9 5.2 5.5 2.8 6.1 6.4 2.6 1.7 4.3

# can we conclude that the median length of pygmy sunfish differs significantly from 3.7 centimeters?
```

```python
l=[5.0, 3.9, 5.2, 5.5, 2.8, 6.1, 6.4 ,2.6 ,1.7, 4.3] 
```

```python
#Hypothesis:
#Null: H0=3.7
#Alternative: H1:=! 3.7
```

```python
from scipy.stats import wilcoxon
import statistics as stat
```

```python
l_median=stat.median(l)
print(l_median)
```

```python
result=wilcoxon(l)
result.statistic,result.pvalue
```

```python
res=wilcoxon(l,alternative='greater')
res.statistic,res.pvalue
```

```python
# Comment: the null hypothesis is rejected.that means the median (3.7) is differ.
```

# Mann Whitny U Test

```python
# code for Mann-Whitney U test 
from scipy.stats import mannwhitneyu 

Placebo =[7, 5, 6, 4, 12] 
New_drug=[3,6,4,2,1] 

# perform mann whitney test 
stat, p_value = mannwhitneyu(Placebo,New_drug) 
print('Statistics=%.2f, p=%.2f' % (stat, p_value)) 
# Level of significance 
alpha = 0.05
# conclusion 
if p_value < alpha: 
	print('Reject Null Hypothesis (Significant difference between two samples)') 
else: 
	print('Do not Reject Null Hypothesis (No significant difference between two samples)')
```
