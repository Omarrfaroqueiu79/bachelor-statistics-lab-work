```python
# Define dependent and independent variables
dependent_variable = 'Common_Mental_Health_Status'  
independent_variables = df.columns[df.columns != dependent_variable]  

# Calculate percentage distribution for each independent variable
percentage_data = {}

for col in independent_variables:
    # Calculate percentage distribution grouped by the dependent variable
    percentages = (df.groupby(col)[dependent_variable]
                   .value_counts(normalize=True)
                   .unstack(fill_value=0) * 100)
    percentage_data[col] = percentages

# Plot bar chart
plt.figure(figsize=(14, 10))

for i, (feature, percentages) in enumerate(percentage_data.items()):
    plt.subplot(6, 3, i + 1)  # Grid of 6 rows and 3 columns
    percentages.plot(kind='bar', stacked=True, ax=plt.gca(), color=['skyblue', 'lightcoral'])  # Add more colors if needed
    plt.title(f'{feature}')
    plt.ylabel('Percentage')
    plt.xlabel('Categories')
    plt.tight_layout()

# Show the plot
plt.show()
```

```python
# Create a Word document
doc = Document()
doc.add_heading('Bar Charts of Independent Variables', level=1)

# Generate bar charts and insert them into Word document
for i, col in enumerate(independent_variables):
    # Calculate percentage distribution
    grouped_data = (df.groupby(col)[dependent_variable]
                    .value_counts(normalize=True)
                    .unstack(fill_value=0) * 100)
    
    # Create bar chart
    plt.figure(figsize=(8, 5))
    grouped_data.plot(kind='bar', stacked=True, color=['skyblue', 'lightcoral', 'lightgreen'], ax=plt.gca())
    plt.title(f'{col}', fontsize=12)
    plt.ylabel('Percentage')
    plt.xlabel(col)
    plt.tight_layout()

    # Save plot to a temporary image file
    image_path = f'temp_plot_{i}.png'
    plt.savefig(image_path, format='png')
    plt.close()

    # Insert chart into Word document
    doc.add_heading(col, level=2)
    doc.add_picture(image_path, width=Inches(6))
    doc.add_paragraph(f"This chart shows the percentage distribution of the dependent variable '{dependent_variable}' across the categories of the independent variable '{col}'.")

# Save the Word document
output_file = 'E:/Bar_Charts_of_Independent_Variables.docx'
doc.save(output_file)

print(f"Bar charts successfully exported to: {output_file}")
```

```python
# Identify independent variables
dependent_var = 'Common_Mental_Health_Status'  
independent_vars = [col for col in df.columns if col != dependent_var]

# Create a Word document
doc = Document()
doc.add_heading('Pie Charts for Independent Variables', level=1)

# Draw pie charts for each independent variable
for var in independent_vars:
    # Calculate the value counts for the current independent variable
    counts = df[var].value_counts()

    # Create the pie chart
    plt.figure(figsize=(6, 6))
    plt.pie(
        counts,
        labels=counts.index,
        autopct='%1.1f%%',
        startangle=140,
        colors=plt.cm.tab20.colors  
    )
    plt.title(f"Distribution of {var}")

    # Save the pie chart as an image
    image_path = f"{var}_distribution.png"
    plt.savefig(image_path, bbox_inches='tight')
    plt.close()

    # Add the pie chart and a description to the Word document
    doc.add_heading(f"Distribution of {var}", level=2)
    doc.add_picture(image_path, width=Inches(4.5))

# Save the Word document
doc.save('E:/Pie_Charts_of_Independent_Variables.docx')

print("Pie charts have been generated and saved to 'Pie_Charts_Independent_Variables.docx'.")
```

```python
# Drop target column and select independent variables
independent_vars = df.drop(columns=['Common_Mental_Health_Status'])  # Adjust the target column name

# Set Seaborn style (optional)
sns.set(style="whitegrid")

# Create a figure for the histogram
plt.figure(figsize=(15, 10))

# Define the number of bins for the histogram
bins = 20

# Define a color palette
colors = sns.color_palette("Set2", n_colors=len(independent_vars.columns))

# Plot a standard histogram for each independent variable
for i, col in enumerate(independent_vars.columns):
    # Plot the histogram for each variable with absolute frequency (no density scaling)
    plt.hist(independent_vars[col], bins=bins, alpha=0.7, label=col,
             color=colors[i % len(colors)], edgecolor='black')

# Add titles and labels
plt.title('Histograms of Independent Variables', fontsize=18, weight='bold')
plt.xlabel('Values', fontsize=14)
plt.ylabel('Frequency', fontsize=14)

# Add a legend
plt.legend(title="Variables", title_fontsize='13', fontsize='12')

# Save the plot to a BytesIO buffer
img_stream = BytesIO()
plt.savefig(img_stream, format='png')  
img_stream.seek(0)  

# Create a Word document
doc = Document()
doc.add_heading('Histograms of Independent Variables', 0)

# Add the combined plot image to the Word document
doc.add_picture(img_stream, width=Inches(6))

# Save the Word document
doc.save('E:/standard_histograms_report.docx')

# Show the plot
plt.show()  
```
