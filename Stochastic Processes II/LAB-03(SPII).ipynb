## Import the given information

```python
# Given values
lambda_rate = 5  # arrival rate (customers per hour)
mu_rate = 8      # service rate (customers per hour)
```

## 03(a)

```python
# Probability that the server is idle
P_idle = 1 - (lambda_rate / mu_rate)
print("Probability that the server is idle: ", P_idle)
```

## 03(b)

```python
# Expected number of customers in the system
L = lambda_rate / (mu_rate - lambda_rate)
print("Expected number of customers in the system: ", L)
```

## 03(c)

```python
# Average time spent in the system
W = 1 / (mu_rate - lambda_rate)
print("Average time spent in the system: ", W)
```

```python
# Print results
print(f"Probability that the server is idle: {P_idle*100:.2f}%")
print(f"Expected number of customers in the system: {L:.4f}")
print(f"Average time a customer spends in the system: {W:.4f} hours ({W*60:.2f} minutes)")
```
