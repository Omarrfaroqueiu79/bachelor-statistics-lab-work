```python
# Load necessary libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Set the number of steps
n_steps = 50
```

```python
# Initialize the starting position
position = np.zeros(n_steps + 1)
```

## 05(a)

```python
# Simulate the random walk
np.random.seed(42)  # For reproducibility

for i in range(1, n_steps + 1):
    # Move left or right with probability 1/2
    move = np.random.choice([-1, 1])
    position[i] = position[i - 1] + move
print("Position:")
print(position)
```

```python
# Plot the position of the particle at each step
plt.plot(range(n_steps + 1), position, marker='o', color='blue', linewidth=2)
plt.title("Random Walk - 50 Steps")
plt.xlabel("Step")
plt.ylabel("Position")
plt.grid(True)
plt.show()
```

## 05(b)

```python
# Final position after 50 steps
final_position = position[-1]
print(f"Final position after 50 steps: {final_position}")
```
