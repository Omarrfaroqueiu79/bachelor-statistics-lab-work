```python
# Import necessary libraries
import numpy as np
import pandas as pd
```

```python
# Define transition matrix P with row and column names
states = ["Sunny", "Cloudy", "Rainy"]
P = pd.DataFrame([[0.7, 0.2, 0.1],
                  [0.3, 0.4, 0.3],
                  [0.2, 0.5, 0.3]], 
                 index=states, columns=states)
print("Transition Matrix:")
print(P)
```

## 01(a)

```python
# Initial state (today is Sunny)
initial_state = np.array([1, 0, 0])  # [Sunny, Cloudy, Rainy]
```

```python
# Probability of being Rainy in 2 days
P_2 = np.linalg.matrix_power(P.values, 2)
prob_rainy_2_days = np.dot(initial_state, P_2)[2]  # Extract Rainy probability
```

```python
# Print the results
print("Probability of being Rainy in 2 days:")
print(f" If today is Sunny, then after 2 days, the probability that it will be Rainy is: {prob_rainy_2_days:.4f}.")
print(f" This means there is a {prob_rainy_2_days*100:.2f}% chance of experiencing Rainy weather after 2 days.")
```

## 01(b)

```python
# Compute steady-state probabilities π (solving πP = π)
A = np.vstack([P.values.T - np.eye(3), np.ones(3)])  # Augmented system
b = np.array([0, 0, 0, 1])  # Constraint sum(π) = 1
steady_state = np.linalg.lstsq(A, b, rcond=None)[0]
```

```python
# Convert steady-state results to a DataFrame
steady_state_df = pd.DataFrame(steady_state, index=states, columns=["Steady-State Probability"])
```

```python
# Print the results
print("\n Long-term (Steady-State) Probabilities:")
print(steady_state_df)
print("\n This means that in the long run, the city will have:")
print(f" - {steady_state[0]*100:.2f}% Sunny days,")
print(f" - {steady_state[1]*100:.2f}% Cloudy days, &")
print(f" - {steady_state[2]*100:.2f}% Rainy days.")
print("These probabilities remain stable over time.")
```
