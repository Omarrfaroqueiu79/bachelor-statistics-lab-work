```python
# Import necessary library
import numpy as np
```

```python
# Set seed for reproducibility
np.random.seed(420)
```

```python
# Given parameters
time_limit = 50
mean_lifetime = 10
lambda_ = 1 / mean_lifetime
```

## 02(a)

```python
# Expected number of replacements
expected_replacements = time_limit / mean_lifetime
print("Expected number of light bulbs replaced in 50 hours:", expected_replacements)
```

## 02(b)

```python
# Simulating the replacement process 
lifetimes = np.random.exponential(scale=mean_lifetime, size=1000)  # Generate a large number of lifetimes
cumulative_time = np.cumsum(lifetimes)  # Compute cumulative sum of lifetimes
```

```python
# Count how many lifetimes fit within the 50-hour period
simulated_replacements = np.sum(cumulative_time < time_limit)
print(f"Simulated number of replacements: {simulated_replacements}")
```
